```python
df=pd.merge(df,df_loc,right_on="WBAN",left_on="STA",how="inner")
```

```python
df.head()
```

```python
df = df.select_dtypes(include=['number'])
```

```python
df = df.dropna()
```

```python
X = df.drop(["MaxTemp", "MAX", "MeanTemp", "MEA", "MinTemp", "MIN"], axis=1)
y = df["MaxTemp"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_tr,X_te,y_tr,y_te = train_test_split(X,y,test_size=0.25,random_state=15)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_tr = scaler.fit_transform(X_tr)
X_te = scaler.transform(X_te)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_tr,y_tr)
```

```python
predicts=model.predict(X_te)
```

```python
print(model.score(X_te,y_te))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model1=RandomForestRegressor(random_state=15)
```

```python
model1.fit(X_tr,y_tr)
```

```python
print(model1.score(X_te,y_te))
```

```python
import joblib
```

```python
joblib.dump(model,"LinearRegressionModel.pkl")
joblib.dump(model1,"RandomForestRegressionModel.pkl")
```

```python
joblib.dump(scaler,"scaler.pkl")
```
